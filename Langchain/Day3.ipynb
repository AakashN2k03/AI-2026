## Streaming And Batch ##

##### Streaming means receiving the model's response piece by piece as it is generated, instead of waiting for the complete response. #####

````python
for chunk in model.stream("Explain AI"):
    print(chunk.content,end="😎",flush=True)
````

##### Batch means sending multiple inputs to the model at once and getting multiple responses. #####

````python
responses = model.batch([
    "What is AI?",
    "What is ML?",
    "What is Deep Learning?"
])

for response in responses:
    print(response)
````

### Batch Execution and `max_concurrency`

By default, `model.batch()` can process multiple inputs **concurrently (in parallel)** rather than always running them one by one.

```text
Without max_concurrency
→ LangChain uses its default concurrency limit.

With max_concurrency = 5
→ A maximum of 5 requests can run at the same time.

With max_concurrency = 1
→ Requests run one by one.

````python
responses = model.batch([
    "What is AI?",
    "What is ML?",
    "What is Deep Learning?"
],
config={
    'max_concurrency':5
})

for response in responses:
    print(response)
````

# Easy Analogy of Temperature,Top-K,Top-P

Imagine you're choosing a student from a class.

## 🌡️ Temperature

Controls how predictable your choice is.

* **Low Temperature** → Always choose the best student.
* **High Temperature** → Might choose someone unexpected.

> 👉 **Controls randomness and creativity.**

---

## 🔝 Top-K

**Top-K = 5**

You can only choose from the **top 5 students**.

> 👉 **Controls the number of choices available.**

---

## 🔵 Top-P

**Top-P = 90%**

You choose students until their combined **probability/confidence** reaches a certain threshold.

The number of students can change depending on their probabilities.

> 👉 **Controls choices based on cumulative probability.**


## Tool ##
##### A tool gives an AI model the ability to interact with external systems and perform actions beyond just generating text. #####
* **`@tool`** → Converts a Python function into a tool that an AI model can understand and use.
* **`bind_tools()`** → Makes those tools available to the AI model so it can decide when to call them.


````python
from langchain.tools import tool

@tool
def get_weather(location:str)->str:
    """Get the weather at a location"""
    return f"It's sunny in {location}"

model_with_tools=model.bind_tools([get_weather])
````

````python
response=model_with_tools.invoke("what's the weather in Kerala")
print(response)

for tool_call in response.tool_calls:
    print(f"Tool: {tool_call['name']}")
    print(f"Tool: {tool_call['args']}")
````

````python
for tool_call in response.tool_calls:
    response=get_weather.invoke(tool_call['args'])
    print(response)
````
