### Messages

Messages are the fundamental units of context for models in LangChain. They represent the input and output of a model and help maintain the state of a conversation when interacting with an LLM.

A message typically contains:

- **Role** – Identifies the message type, such as `system`, `user`, or `assistant`.
- **Content** – Represents the actual message content, such as text, images, audio, or documents.
- **Metadata** – Optional additional information, such as response details, message IDs, and token usage.

### SystemMessage

A `SystemMessage` gives instructions to the AI about **how it should behave**.

**Example:**

```python
SystemMessage(
    content="You are a helpful Python teacher. Explain everything simply."
)
```

### HumanMessage

A `HumanMessage` represents a **message sent by the user**.

**Example:**

```python
HumanMessage(
    content="What is a Python function?"
)
```

### AIMessage

An `AIMessage` represents the **response generated by the AI model**.
AIMessage stores previous AI responses for conversation context.

**Example:**

```python
AIMessage(
    content="A function is a reusable block of code that performs a specific task."
)
```

### ToolMessage

A `ToolMessage` contains the **result returned by a tool** and sends that result back to the AI model.


**Example:**

```python
ToolMessage(
    content="The weather in Chennai is 32°C and sunny.",
    tool_call_id="123"
)
```

### Simple Flow

```text
SystemMessage → Tells the AI how to behave
HumanMessage  → User asks a question
AIMessage     → AI generates a response
ToolMessage   → Returns a tool's result back to the AI
``` 

````python
import os
from dotenv import load_dotenv
load_dotenv()

api_key=os.getenv("GROQ_API_KEY")
````

````python
from langchain.chat_models import init_chat_model

model=init_chat_model("groq:qwen/qwen3.6-27b")
````

````python

from langchain.messages import HumanMessage,SystemMessage,AIMessage
messages=[
    SystemMessage("You are a poetry expert"),
    HumanMessage("Write a poem on artificial intelligence")
]

response=model.invoke(messages)
print(response.content)
````

````python
## Message Metadata
human_msg = HumanMessage(
    content="Hello!",
    name="alice",  # Optional: identify different users
    id="msg_123",  # Optional: unique identifier for tracing
)

response=model.invoke([human_msg])
print(response.content)
````

````python
from langchain.messages import HumanMessage, SystemMessage, AIMessage

response = model.invoke([
    SystemMessage("You are a Maths Tutor"),
    HumanMessage("What is the Pythagorean Theorem?"),
    AIMessage("It's a mathematical formula relating the sides of a right-angled triangle."),
    HumanMessage("Great! Then what is the hypotenuse?")
])

print(response.content)
````

````python
response.usage_metadata
````

````python
from langchain.messages import ToolMessage

ai_message = AIMessage(
    content=[],
    tool_calls=[{
        "name": "get_weather",
        "args": {"location": "San Francisco"},
        "id": "call_123"
        #The id is not used to decide which tool to call. It is used to match the tool result with the specific tool call.
    }]
)

# Execute tool and create result message
weather_result = "Sunny, 72°F"
tool_message = ToolMessage(
    content=weather_result,
    tool_call_id="call_123"  # Must match the call ID
)

# Continue conversation
messages = [
    HumanMessage("What's the weather in San Francisco?"),
    ai_message,  # Model's tool call
    tool_message,  # Tool execution result
]
response = model.invoke(messages)  # Model processes the result
print(response.content)
````
