## 1. `temperature`

Controls how deterministic the model is.

- `0` → more predictable / deterministic
- `0.2` → slightly varied
- `0.7` → more creative
- Higher → more randomness

For agents, coding, SQL and tool calling, `temperature=0` is generally the right default.

### NOTE: 
- Total tokens = input tokens + output tokens.
- Output tokens = tokens generated by the model in its response.

```python
from  langchain_google_genai import ChatGoogleGenerativeAI
model = ChatGoogleGenerativeAI(
    model="gemini-3.6-flash",
    temperature=0.8,
)

response = model.invoke("Give me one short tagline for a coffee shop.")
print(response.content)
```

```python
# Compare determinism: temperature=0 vs temperature=0.9
prompt = "Give me one short tagline for a coffee shop."

for temp in (0, 0.9):
    m = ChatGoogleGenerativeAI(model="gemini-3.6-flash", temperature=temp)
    print(f"--- temperature={temp} ---")
    for _ in range(2): # _ means the loop variable is not needed; we only want to repeat the loop.
        print(m.invoke(prompt).content)
```

## 2. `max_tokens`

Controls the maximum length of the generated response.

This is a **generation limit**, not the size of the input/context.

```python
model = ChatGoogleGenerativeAI(
    model="gemini-3.6-flash",
    max_tokens=1000,
)

response = model.invoke("Explain vector databases.")
print(response.content)
```

## 3. `timeout`

How long LangChain waits for the API call (in seconds).
Timeout controls how long your application waits for the model's response before giving up.

Useful when you're building an agent that may take longer because of tool calls.

```python
model = ChatGoogleGenerativeAI(
    model="gemini-3.6-flash",
    timeout=60,
)

print(model.invoke("Say hello.").content)

# Using .content gives you only the model's generated text, while without .content you get the complete response object.
```

## 4. `max_retries`

Number of times LangChain retries certain temporary API failures.

For example, if there is a temporary rate-limit or server issue, LangChain can retry
instead of raising immediately.

```python
model = ChatGoogleGenerativeAI(
    model="gemini-3.6-flash",
    max_retries=2,
)

print(model.invoke("Say hello.").content)
```

## 5. `top_p`

top_p controls how many of the most likely token choices the model can consider when generating a response.
considers tokens until their combined probability reaches a limit, e.g. 90%.
You usually don't need to change this if you're already controlling `temperature`.

```python
model = ChatGoogleGenerativeAI(
    model="gemini-3.6-flash",
    top_p=0.9,
)

print(model.invoke("Name three uses for a paperclip.").content)
```

## 6. `top_k`

top_k controls how many of the most likely token choices the model can consider when generating a response. It considers a fixed number of the most likely tokens, for example, the top 40.

Again, this is usually something you can leave at its default value.

```python
model = ChatGoogleGenerativeAI(
    model="gemini-3.6-flash",
    top_k=40,
)

print(model.invoke("Name three uses for a paperclip.").content)
```

## 7. `streaming`

Controls whether output is streamed.

Instead of waiting for the complete response, you can receive chunks as they're generated.

```python
model = ChatGoogleGenerativeAI(
    model="gemini-2.5-flash",
    streaming=True,
)

for chunk in model.stream("Write a short paragraph about the sea."):
    print(chunk.content, end="", flush=True)

# end="" → prevents print() from adding a new line after each chunk, so the streamed text appears continuously.
# flush=True → immediately displays each chunk instead of waiting for Python's output buffer.
```

## 8. `response_mime_type`

Useful when you want structured output, especially JSON.

You can then combine this with LangChain structured-output features.

```python
model = ChatGoogleGenerativeAI(
    model="gemini-3.5-flash-lite",
    response_mime_type="application/json",
)

response = model.invoke(
    "List 3 programming languages with a `name` and `year` field. Return JSON only."
)
print(response.content)
```

```python
model = ChatGoogleGenerativeAI(
    model="gemini-3.5-flash-lite",
    response_mime_type="text/plain",
)

response = model.invoke(
    "List 3 programming languages with a `name` and `year` field. Return JSON only."
)
print(response.content)
```
